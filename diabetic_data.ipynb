```python
#Import Libraries
import pandas as pd      # For data manipulation
import numpy as np       # For numerical operations
import matplotlib.pyplot as plt  # For basic plots
import seaborn as sns    # For advanced statistical visualizations

sns.set(style="whitegrid")

#Load dataset
url = "https://raw.githubusercontent.com/niteen11/DataAnalyticsAcademy/master/Python/dataset_diabetes/diabetic_data.csv"
df = pd.read_csv(url)
df.head()

```

```python
print('Shape (rows, columns):', df.shape)

print('\nColumn names:')
print(df.columns)

print('\nData types and non-null counts:')
df.info()

print('\nSummary statistics (numeric columns):')
df.describe()
```

```python
#Missing Values & Unique Values
print('Missing values per column:')
print(df.isna().sum())

print('\nPercentage of missing values per column:')
print((df.isna().mean() * 100).round(2))

print('\nNumber of unique values per column:')
print(df.nunique())
```

```python
df.columns = df.columns.str.title()
df.head()
```

```python
#Data Cleaning Task
# Check unique values in gender
print("Unique values in 'Gender' before cleaning:", df['Gender'].unique())

# Count patients with missing or invalid gender
missing_gender_count = df['Gender'].isna().sum()
invalid_gender_count = (df['Gender'] == 'Unknown/Invalid').sum()

print("Number of patients with missing gender:", missing_gender_count)
print("Number of patients with invalid gender:", invalid_gender_count)

# Optional: total without valid gender
total_no_gender = missing_gender_count + invalid_gender_count
print("Total patients without valid gender:", total_no_gender)


```

A small number of patient records contained missing or invalid gender values and were identified during the data cleaning process to ensure accurate analysis.

```python
#Descriptive Statistics
# Compare 'time_in_hospital' between Male and Female
group_stats = df.groupby('Gender')['Time_In_Hospital'].describe()
print("\n'Time_in_Hospital' by Gender:\n")
print(group_stats)
```

```python
#Descriptive Statistics
# Compare 'race' distribution between Male and Female
race_stats = pd.crosstab(df['Race'], df['Gender'], normalize='columns') * 100
print("\nRace Distribution (%) by Gender:\n")
print(race_stats)

#absolute counts as well
race_counts = pd.crosstab(df['Race'], df['Gender'])
print("\nRace counts by Gender:\n")
print(race_counts)
```

```python
#Visualization
race_counts.plot(kind='bar', stacked=True, figsize=(8,5), color=['skyblue','pink'])
plt.title("Race Distribution by Gender")
plt.xlabel("Race")
plt.ylabel("Number of Patients")
plt.legend(title="Gender")
plt.show()
```

Male and Female patients show similar racial distributions, with most patients identifying as White. Minor differences exist among other racial groups, but overall patterns are comparable.

```python
#distribution
plt.figure(figsize=(8,5))
sns.histplot(df[df['Gender']=='Male']['Time_In_Hospital'], color='blue', label='Male', kde=True, bins=15)
sns.histplot(df[df['Gender']=='Female']['Time_In_Hospital'], color='pink', label='Female', kde=True, bins=15)
plt.title("Distribution of Time in Hospital by Gender")
plt.xlabel("Time in Hospital (days)")
plt.ylabel("Count")
plt.legend()
plt.show()
```

On average, Male and Female patients spend a similar amount of time in the hospital. However, the boxplot shows that Male patients have slightly more variation,with some staying longer than most Female patients.

```python
#Descriptive Statistics: Time in Hospital by Race
race_time_stats = df.groupby('Race')['Time_In_Hospital'].describe()
print("Time in Hospital by Race:\n")
print(race_time_stats)

#Visualization: Boxplot
plt.figure(figsize=(10,6))
sns.boxplot(x='Race', y='Time_In_Hospital', data=df, palette='Set2')
plt.title("Time in Hospital by Race")
plt.xlabel("Race")
plt.ylabel("Time in Hospital (days)")
plt.show()

#Optional: Barplot of Average Time in Hospital by Race
plt.figure(figsize=(8,5))
avg_time = df.groupby('Race')['Time_In_Hospital'].mean().sort_values()
sns.barplot(x=avg_time.index, y=avg_time.values, palette='Set3')
plt.title("Average Time in Hospital by Race")
plt.xlabel("Race")
plt.ylabel("Average Time in Hospital (days)")
plt.show()
```

Patients’ length of stay varies slightly across racial groups. On average, Black and Other race patients tend to stay in the hospital a bit longer than White and Asian patients, though the differences are modest. Overall, race shows some influence on hospital stay, but no extreme disparities are evident.

```python
# Count how many times each patient appears in the dataset
admission_counts = df['Patient_Nbr'].value_counts()

# Display top 10 patients with most admissions
print(admission_counts.head(10))
```

Patients with multiple hospital admissions may indicate difficulty in managing diabetes outside the hospital setting. While the dataset does not identify specific causes, the pattern suggests a need for enhanced continuity of care to help prevent repeat admissions.

```python
#Count admissions per patient
Admission_Counts = df['Patient_Nbr'].value_counts()

#Counts how many patients fall into each admission frequency
admission_frequency = Admission_Counts.value_counts().sort_index()

#Visualization
#x-axis: How many times a patient was admitted (1, 2, 3, …).
#Y-axis: How many patients fall into each category.

plt.figure(figsize=(8,5))
sns.barplot(x=admission_frequency.index, y=admission_frequency.values, palette='viridis')
plt.title("Number of Patients by Admission Frequency")
plt.xlabel("Number of Admissions")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.show()
```

```python
#Create categories: single vs multiple admissions
single_admission = (Admission_Counts == 1).sum()
multiple_admissions = (Admission_Counts > 1).sum()

labels = ['Single Admission', 'Multiple Admissions']
sizes = [single_admission, multiple_admissions]

#Pie Chart
plt.figure(figsize=(6,6))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Percentage of Patients with Single vs Multiple Admissions")
plt.axis('Equal')  # Ensures pie is a circle
plt.show()
```

Most patients in the dataset were admitted only once, while a smaller but notable portion experienced multiple hospital admissions. This highlights the chronic nature of diabetes and the importance of effective outpatient management to reduce readmissions.
